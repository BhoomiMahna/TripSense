```python
import pandas as pd
df=pd.read_csv('travel_data.csv')
df.head(30)
```

```python
df.info()
```

```python
df.isnull().sum()
df.describe()
```

```python
df.isnull().sum()
```

```python
df['TravelCostINR']=df['TravelCostINR'].fillna(df['TravelCostINR'].median())

if not df['NetworkQuality'].mode().empty:
      df['NetworkQuality']=df['NetworkQuality'].fillna(df['NetworkQuality'].mode()[0])
else:
     df['NetworkQuality'] = df['NetworkQuality'].fillna('Unknown')
```

```python
df.drop_duplicates(inplace=True)
```

```python
numeric_cols=df.select_dtypes(include='number').columns
negatives=df[numeric_cols]<0
neg_counts=negatives.sum()
neg_counts[neg_counts>0]
```

```python
df[numeric_cols]=df[numeric_cols].abs()
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for col in ['Gender','Destination','Category','State','District','NetworkQuality','PermitsRequired']:
  df[col]=le.fit_transform(df[col])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
trip_freq=df.groupby('CustomerID').size().reset_index(name='trip_frequency_per_year')
```

```python
customer_agg = df.groupby('CustomerID').agg({
    'Age':'first',
    'Gender':'first',
    'NatureScore':'mean',
    'CultureScore':'mean',
    'AdventureScore':'mean',
    'SpiritualScore':'mean',
    'BookingLeadDays':'mean',
    'DurationDays':'mean',
    'FamilySize':'mean',
    'InternetSpeedMbps':'mean',
    'NetworkQuality':'mean',
    'PermitsRequired':'mean',
    'Destination':'nunique',
    'Category':'nunique'
}).reset_index()


customer_agg.rename(columns={'Destination':'destination_diversity','Category':'category_diversity'}, inplace=True)
```

```python

customer_agg = customer_agg.merge(trip_freq, on='CustomerID', how='left')

customer_agg.head()

```

```python
print(customer_agg.columns.tolist())
```

these features can be used ofr clustering so making them

```python

customer_agg['adventure_pref_score'] = customer_agg['AdventureScore'] * customer_agg['DurationDays']
customer_agg['culture_pref_score'] = customer_agg['CultureScore'] * customer_agg['BookingLeadDays']
customer_agg['nature_pref_score'] = customer_agg['NatureScore'] * customer_agg['DurationDays']
customer_agg['spiritual_pref_score'] = customer_agg['SpiritualScore'] * customer_agg['DurationDays']
customer_agg['engagement_score'] = customer_agg['destination_diversity'] * customer_agg['DurationDays']
customer_agg['avg_booking_value_inr'] = (
    customer_agg['DurationDays'] * 3000 * customer_agg['FamilySize']
)
customer_agg['international_trip_ratio'] = (
    (customer_agg['PermitsRequired'] == 1).astype(int)
)
customer_agg['luxury_hotel_ratio'] = (
    (customer_agg['NatureScore'] +
     customer_agg['CultureScore'] +
     customer_agg['SpiritualScore']) / 30
)
customer_agg['budget_booking_ratio'] = (
    customer_agg['AdventureScore'] / 10
)
customer_agg['price_sensitivity_score'] = (
    customer_agg['FamilySize'] / (customer_agg['BookingLeadDays'] + 1)
)
customer_agg['weekend_trip_ratio'] = (
    (customer_agg['DurationDays'] <= 3).astype(int)
)
customer_agg['last_minute_search_ratio'] = (
    (customer_agg['BookingLeadDays'] <= 5).astype(int)
)
customer_agg['cancellation_rate'] = (
    1 / (customer_agg['engagement_score'] + 1)
)
customer_agg['loyalty_points_balance'] = (
    customer_agg['trip_frequency_per_year'] *
    customer_agg['engagement_score'] * 10
)
customer_agg['mobile_booking_ratio'] = (
    customer_agg['InternetSpeedMbps'] / 100
)


```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()

numeric = [
    'Age',
    'Gender',
    'NatureScore',
    'CultureScore',
    'AdventureScore',
    'SpiritualScore',
    'BookingLeadDays',
    'DurationDays',
    'FamilySize',
    'InternetSpeedMbps',
    'NetworkQuality',
    'PermitsRequired',
    'destination_diversity',
    'category_diversity',
    'trip_frequency_per_year',
    'adventure_pref_score',
    'culture_pref_score',
    'nature_pref_score',
    'spiritual_pref_score',
    'engagement_score',
    'avg_booking_value_inr',
    'international_trip_ratio',
    'luxury_hotel_ratio',
    'budget_booking_ratio',
    'price_sensitivity_score',
    'weekend_trip_ratio',
    'last_minute_search_ratio',
    'cancellation_rate',
    'loyalty_points_balance',
    'mobile_booking_ratio'
]

customer_agg[numeric]=scaler.fit_transform(customer_agg[numeric])
```

```python
import numpy as np
np.random.seed(42)
customer_agg['target_booking'] = np.random.choice([0,1], size=len(customer_agg), p=[0.3,0.7])
customer_agg['target_response'] = np.random.choice([0,1], size=len(customer_agg), p=[0.4,0.6])

```

```python
augmented_data=customer_agg.copy()
```

```python
numeric_cols = [
    'Age',
    'Gender',
    'NatureScore',
    'CultureScore',
    'AdventureScore',
    'SpiritualScore',
    'BookingLeadDays',
    'DurationDays',
    'FamilySize',
    'InternetSpeedMbps',
    'NetworkQuality',
    'PermitsRequired',
    'destination_diversity',
    'category_diversity',
    'trip_frequency_per_year',
    'adventure_pref_score',
    'culture_pref_score',
    'nature_pref_score',
    'spiritual_pref_score',
    'engagement_score',
    'avg_booking_value_inr',
    'international_trip_ratio',
    'luxury_hotel_ratio',
    'budget_booking_ratio',
    'price_sensitivity_score',
    'weekend_trip_ratio',
    'last_minute_search_ratio',
    'cancellation_rate',
    'loyalty_points_balance',
    'mobile_booking_ratio'
]


np.random.seed(42)
noise_factor=0.05
synthetic_rows=augmented_data.copy()
for col in numeric_cols:
    noise = np.random.normal(0, noise_factor, size=len(synthetic_rows))
    synthetic_rows[col] = synthetic_rows[col] * (1 + noise)
synthetic_rows[numeric_cols] = synthetic_rows[numeric_cols].clip(0, 1)
```

```python
augmented_data = pd.concat([augmented_data, synthetic_rows], ignore_index=True)
print("Original rows:", len(customer_agg))
print("Augmented rows:", len(augmented_data))
```

```python
augmented_data = augmented_data.sample(frac=1, random_state=42).reset_index(drop=True)
```
